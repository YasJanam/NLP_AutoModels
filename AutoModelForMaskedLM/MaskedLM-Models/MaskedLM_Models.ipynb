#### **example 1**

```python
from transformers import AutoTokenizer, AutoModelForMaskedLM, pipeline

# مدل‌های خانواده BERT برای Masked LM
models = [
    "bert-base-uncased",        # BERT اصلی
    "roberta-base",             # RoBERTa
    "distilbert-base-uncased",  # DistilBERT
    "albert-base-v2",           # ALBERT
    "google/electra-base-generator"  # ELECTRA (generator part)
]

text = "The capital of France is [MASK]."
text_roberta = "The capital of France is <mask>." # Use <mask> for RoBERTa

count = 1
for model_name in models:
    print(f"\n{count}. 🟥 Model: {model_name} 🟥")
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModelForMaskedLM.from_pretrained(model_name)

    nlp_fill = pipeline("fill-mask", model=model, tokenizer=tokenizer)

    if model_name == "roberta-base":
        results = nlp_fill(text_roberta)
    else:
        results = nlp_fill(text)

    print("--------------------")

    for r in results[:3]:  # سه پیشنهاد اول
        print(f"⭐  {r['sequence']} (prob={r['score']:.4f})")

    print("--------------------")

    count +=1
```

```python
from transformers import AutoTokenizer, AutoModelForMaskedLM, pipeline

dict_models = {
    "BERT":"bert-base-uncased",
    "RoBERTa":"roberta-base",
    "DistilBERT":"distilbert-base-uncased",
    "ALBERT":"albert-base-v2",
    "ELECTRA":"google/electra-base-generator",
    "SpanBERT":"SpanBERT/spanbert-large-cased",
    "MiniLM":"microsoft/MiniLM-L12-H384-uncased"
    }

count = 1
for name, model_name in dict_models.items():
    print("______________________________________")
    print(f"\n⭐⭐{count}.{name}-Model⭐⭐")
    print(f"\n 🟥 Model-name : {model_name}🟥\n ")

    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModelForMaskedLM.from_pretrained(model_name)

    print("--------------------")
    print(f"🔹 model-type -> {model.config.model_type}")
    print(f"🔹 vocab-size -> {model.config.vocab_size}")
    print(f"🔹 hidden-size -> {model.config.hidden_size}")
    print(f"🔹 num-hidden-layers -> {model.config.num_hidden_layers}")
    print(f"🔹 neum-attention-heads -> {model.config.num_attention_heads}")
    print("--------------------")

    count +=1
```
